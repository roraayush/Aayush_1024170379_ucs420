## Q1. Lists

```
# Roll number and digit extraction
roll_number = "1024170379"
digits = [int(d) for d in roll_number]
print("Digits of roll number:", digits)

# Multiply each digit by 10 to form list L
L = [d * 10 for d in digits]
print("L =", L)
```

**i. Print L**

```
print(L)
```

**ii. Add two numbers — one with append(), one with insert()**

```
# append() adds an element to the END of the list
L.append(55)
print("After append(55):", L)
# insert(position, value) adds an element at a SPECIFIC index, shifting later elements right
L.insert(3, 77)
print("After insert(3, 77):", L)
```

**iii. Remove two elements — one with remove(), one with pop()**

```
# remove(value) deletes the first matching VALUE (raises error if not found)
L.remove(77)
print("After remove(77):", L)
# pop() removes and returns the LAST element by default (or by index if given)
popped = L.pop()
print(f"After pop() (removed {popped}):", L)
```

**iv. Sort ascending, then descending**

```
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)
```

**v. Slicing — first three and last three elements**

```
print("First three:", L[:3])
print("Last three:", L[-3:])
```

**vi. List comprehension — elements greater than the average**

```
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print("Average of L:", avg)
print("Elements greater than average:", above_avg)
```

## Q2. Tuples

```
# First 8 values from list L (as it stood right after Q1)
scores = tuple(L[:8])
print("scores =", scores)
```

**i. Highest score & its index; lowest score & its frequency**

```
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)

print(f"Highest score: {highest} at index {highest_index}")
print(f"Lowest score: {lowest}, occurs {lowest_count} time(s)")
```

**ii. Reverse the tuple (as a list)**

```
reversed_scores = list(scores[::-1])
print("Reversed scores:", reversed_scores)
# Tuples are immutable, so there's no in-place .reverse() method like lists have;
# reversing must produce a NEW sequence (here, a list) instead of modifying scores itself.
```

**iii. Ask user for a score, find its first occurrence index**

```
user_score = int(input("Enter a score to search for: "))
if user_score in scores:
    print(f"{user_score} found at index {scores.index(user_score)}")
else:
    print(f"{user_score} is not present in scores")
```

**iv. Attempt to modify a tuple element (capture the error)**

```
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
# Tuples are immutable — once created, their elements cannot be reassigned,
# unlike lists, which allow item assignment because they are mutable.
```

**v. Unpack into first, second, and the rest**

```
first_score, second_score, *remaining_scores = scores
print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

## Q3. Random Numbers

```
import random

roll_number_int = 1024170379
random.seed(roll_number_int)

random_list = [random.randint(100, 900) for _ in range(100)]
print(random_list)
```

**ii & iii. Count odd and even numbers**

```
odd_numbers = [x for x in random_list if x % 2 != 0]
even_numbers = [x for x in random_list if x % 2 == 0]

print("Count of odd numbers:", len(odd_numbers))
print("Count of even numbers:", len(even_numbers))
```

**iv. Count and list prime numbers**

```
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_list if is_prime(x)]
print("Count of prime numbers:", len(prime_numbers))
print("Prime numbers found:", prime_numbers)
```

**v. Most frequently occurring number**

```
from collections import Counter

counts = Counter(random_list)
most_common_value, most_common_freq = counts.most_common(1)[0]
print(f"Most frequent number: {most_common_value}, occurs {most_common_freq} time(s)")
```

## Q4. Sets

```
# Using the original 8 digits from Q1 (before multiplying by 10)
digits_8 = digits[:8]
print("First 8 digits:", digits_8)

A = {d * 7 for d in digits_8}
B = {d * 9 for d in digits_8}

print("Set A (digit x 7):", A)
print("Set B (digit x 9):", B)
```

**vi. Union**

```
union_AB = A.union(B)
print("A union B:", union_AB)
```

**vii. Intersection**

```
intersection_AB = A.intersection(B)
print("A intersection B:", intersection_AB)
```

**viii. A − B and B − A**

```
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
# difference() gives elements in ONE set but not the other (direction matters: A-B != B-A),
# while symmetric_difference() combines both directions into a single set of all non-shared elements.
```

**ix. Symmetric difference**

```
sym_diff_AB = A.symmetric_difference(B)
print("Symmetric difference of A and B:", sym_diff_AB)
```

**x. Subset / superset check**

```
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

**xi. Discard a user-input value from A**

```
X = int(input("Enter a value to discard from set A: "))
A.discard(X)
print("Set A after discard:", A)
# discard() does nothing if the value isn't present, whereas remove() raises a KeyError —
# so discard() is safer when you're not sure the value exists in the set.
```

## Q5. Dictionaries

```
my_dict = {
    "name": "Aayush",
    "roll_no": "1024170379",
    "branch": "COPC",
    "age": 20,
    "city": "Panipat"
}
print(my_dict)
```

**i. Rename "city" to "location" (using pop())**

```
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

**ii. Add "cgpa" key**

```
my_dict["cgpa"] = 7.76
print(my_dict)
```

**iii. Increase "age" by 1**

```
my_dict["age"] += 1
print(my_dict)
```

**iv. Delete "branch" using pop() and del (in two separate copies)**

```
dict_copy1 = my_dict.copy()
dict_copy2 = my_dict.copy()

popped_value = dict_copy1.pop("branch")
print("pop() returned:", popped_value)
print("dict_copy1 after pop():", dict_copy1)

del dict_copy2["branch"]
print("dict_copy2 after del:", dict_copy2)
# pop() returns the value of the removed key (useful if you need it), while del removes
# the key-value pair directly and returns nothing.
```

**v. Iterate with .items()**

```
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

**vi. Safely check for "email" key**

```
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Key 'email' does not exist in the dictionary")
```

**vii. Merge with friend_dict using {**dict1, **dict2}**

```
friend_dict = {
    "name": "Aryan",
    "roll_no": "1024170555",
    "branch": "COPC",
    "age": 20,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
# When both dictionaries share a key, the value from the SECOND dictionary (friend_dict)
# overwrites the first, since later keys in the unpacking take precedence.
```

**viii. Dictionary comprehension — only string-valued pairs**

```
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_only_dict)
```
