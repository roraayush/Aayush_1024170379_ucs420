Assignment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.

```python
n = int(input("Enter a number: "))
nums = []
for i in range(1, n+1):
  if i%7==0 and i%9==0:
    nums.append(i)

print(f"The sum of numbers divisible by 7 and 9 from 1 to {n} is {sum(nums)}")
```

Assignment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.

```python
n = int(input("Enter a number: "))
added = 0

for i in range(2, n+1):
  isPrime = True
  for j in range(2, i):
    if i%j == 0:
      isPrime = False
      break
  if isPrime:
    added += i

print(added)
```

Assignment 6.1: WAP using function that add all odd numbers from 1 to n, n is given by the user.

```python
def add_odd_num(n):
  added = 0
  for i in range(1, n+1):
    if i%2 != 0:
      added += i
  return added

n = int(input("Enter a number: "))
print(add_odd_num(n))
```

Assignment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.

```python
def add_all_primes(n):
  added = 0
  for i in range(2, n+1):
    isPrime = True
    for j in range(2, i):
      if i%j == 0:
        isPrime = False
        break
    if isPrime:
      added += i
  return added

n = int(input("Enter a number: "))
print(add_all_primes(n))
```
